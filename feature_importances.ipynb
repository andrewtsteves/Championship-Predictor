```python
import models_best_params as M
import pandas as pd
from sklearn.inspection import permutation_importance
import matplotlib.pyplot as plt
```

```python
lr_model = M.lr.named_steps['logisticregression']
coefficients = lr_model.coef_[0]

features = pd.DataFrame({'Features': M.X_train.columns,
                         'Coefficients': coefficients}).sort_values(by='Coefficients', ascending=False).round(3)

print(features)
```

```python
perms = permutation_importance(M.svc, M.X_test, M.y_test)

features = pd.DataFrame({'Features': M.X_test.columns,
                         'Importances': perms.importances_mean}).sort_values(by='Importances', ascending=False).round(3)

print(features)
```

```python
dt_model = M.dt.named_steps['decisiontreeclassifier']
importances = dt_model.feature_importances_

features = pd.DataFrame({'Features': M.X_train.columns,
                         'Importances': importances}).sort_values(by = 'Importances', ascending = False).round(2)
print(features)
```

```python
rf_model = M.rf.named_steps['randomforestclassifier']
importances = rf_model.feature_importances_

features = pd.DataFrame({'Features': M.X_train.columns,
                         'Importances': importances}).sort_values(by = 'Importances', ascending = False).round(2)
print(features)
```
